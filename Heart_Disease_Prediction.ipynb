## 3. Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(12, 5))

# Target distribution
plt.subplot(1, 2, 1)
sns.countplot(x='target', data=df, palette=['#10b981', '#ef4444'])
plt.title('Target Class Distribution (0=Healthy, 1=Heart Disease)')
plt.xlabel('Diagnosis Target')

# Age vs Target
plt.subplot(1, 2, 2)
sns.histplot(data=df, x='age', hue='target', kde=True, palette=['#10b981', '#ef4444'], bins=20)
plt.title('Age Distribution by Heart Disease Status')
plt.xlabel('Patient Age')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
corr = df.corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Clinical Feature Correlation Heatmap')
plt.show()
```

## 6. Real-Time Patient Risk Inference Prototype

```python
# Sample Patient Profile (High Risk)
sample_patient = np.array([[
    62,   # age
    1,    # sex (male)
    0,    # cp (typical angina)
    155,  # resting BP
    280,  # cholesterol
    1,    # fasting blood sugar > 120
    1,    # restecg (ST-T abnormality)
    110,  # max heart rate
    1,    # exercise angina
    2.6,  # ST depression oldpeak
    2,    # ST slope
    2,    # ca major vessels
    3     # thal
]])

model = bundle['best_model']
scaler = bundle['scaler']
scaled_patient = scaler.transform(sample_patient)

prob = model.predict_proba(scaled_patient)[0, 1]
prediction = "LIKELY" if prob >= 0.50 else "NOT LIKELY"

print(f"Prediction: Heart disease is {prediction}")
print(f"Calculated Risk Probability: {prob * 100:.2f}%")
```
